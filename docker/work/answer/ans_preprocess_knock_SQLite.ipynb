## データベース構造の確認

---
> C-001: 顧客テーブル（customer）から全項目の先頭5件を表示せよ。

```python
%%sql
SELECT *
FROM customer
LIMIT 5;
```

---
> C-002: 顧客テーブル（customer）から顧客ID（customer_id）、顧客名（customer_name）、性別（gender）の3列のみを10件表示せよ。

```python
%%sql
SELECT customer_id, customer_name, gender
FROM customer
LIMIT 10;
```

---
> C-003: 顧客テーブル（customer）から女性（gender='女性'）の顧客を5件表示せよ。

```python
%%sql
SELECT *
FROM customer
WHERE gender = '女性'
LIMIT 5;
```

---
> C-004: 新しい顧客を顧客テーブル（customer）に追加せよ。

```python
%%sql
INSERT INTO customer (customer_id, customer_name, gender_cd, gender, age)
VALUES ('CS999999999999', '山田 太郎', '9', '不明', 30);
```

---
> C-005: 追加した顧客（customer_id='CS999999999999'）が正しく登録されたか確認せよ。

```python
%%sql
SELECT *
FROM customer
WHERE customer_id = 'CS999999999999';
```

---
> C-006: 追加した顧客（customer_id='CS999999999999'）の年齢（age）を35に更新せよ。

```python
%%sql
UPDATE customer
SET age = 35
WHERE customer_id = 'CS999999999999';
```

---
> C-007: 更新した顧客（customer_id='CS999999999999'）の情報を表示して、年齢が35に更新されたことを確認せよ。

```python
%%sql
SELECT customer_id, customer_name, age
FROM customer
WHERE customer_id = 'CS999999999999';
```

---
> C-008: 顧客テーブル（customer）の総レコード数を取得せよ。

```python
%%sql
SELECT COUNT(*) AS total_customers
FROM customer;
```

---
> C-009: 複数の顧客を一度に追加せよ。

```python
%%sql
INSERT INTO customer (customer_id, customer_name, gender_cd, gender, age)
VALUES 
    ('CS999999999998', '佐藤 花子', '1', '女性', 25),
    ('CS999999999997', '鈴木 次郎', '0', '男性', 40);
```

---
> C-010: テストで追加した顧客（customer_idが'CS99999999'で始まる）を全て削除せよ。

```python
%%sql
DELETE FROM customer
WHERE customer_id LIKE 'CS99999999%';
```

# データ加工100本ノック（既存の問題）【解答】

## 注意
以下の問題は元々PostgreSQL用に作られています。
SQLiteでも動作しますが、一部の関数や構文が異なる場合があります。

---
> S-001: レシート明細データ（receipt）から全項目の先頭10件を表示し、どのようなデータを保有しているか目視で確認せよ。

```python
%%sql
SELECT *
FROM receipt
LIMIT 10;
```

---
> S-002: レシート明細データ（receipt）から売上年月日（sales_ymd）、顧客ID（customer_id）、商品コード（product_cd）、売上金額（amount）の順に列を指定し、10件表示せよ。

```python
%%sql
SELECT sales_ymd, customer_id, product_cd, amount
FROM receipt
LIMIT 10;
```

---
> S-003: レシート明細データ（receipt）から売上年月日（sales_ymd）、顧客ID（customer_id）、商品コード（product_cd）、売上金額（amount）の順に列を指定し、10件表示せよ。ただし、sales_ymdをsales_dateに項目名を変更して表示すること。

```python
%%sql
SELECT sales_ymd AS sales_date, customer_id, product_cd, amount
FROM receipt
LIMIT 10;
```
