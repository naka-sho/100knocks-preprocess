## はじめに
- このノートブックはSQLiteを使用したCRUD（作成・読込・更新・削除）操作の演習です
- 初めに以下のセルを実行してデータベース接続を設定してください
- セルに %%sql と記載することでSQLを発行することができます
- SQLiteデータベースファイル（dsdojo.db）は `setup_sqlite.py` スクリプトで作成できます

```python
%load_ext sql
import sqlite3
from pathlib import Path

# SQLite database path
db_path = Path('dsdojo.db')

# Check if database exists
if not db_path.exists():
    print('データベースが見つかりません。setup_sqlite.pyを実行してください。')
else:
    # Connect to SQLite database
    %sql sqlite:///dsdojo.db
    print(f'✓ SQLiteデータベースに接続しました: {db_path.absolute()}')
```
